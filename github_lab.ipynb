## Problem 1: Create a GitHub Account & Your First Repository

A **repository** (or *repo*) is a project folder on GitHub. It holds all your files — notebooks, data, and documentation — and keeps a full history of every change you make.

---

## Problem 2: Write a README with Markdown

The **README.md** is the front page of your repo — the first thing anyone sees. It's written in **Markdown**, a lightweight formatting language that turns plain text symbols into formatted headings, lists, and more.

---
### 📌 Example README

Here is a clean, minimal template:

```markdown
# Project Title

A one-sentence description of what this project is about.

## Dataset

- **Source:** Where the data came from (e.g., Kaggle, UCI, class portal)
- **File:** `my_data.csv`
- **Rows:** 500  
- **Key columns:** age, income, target

## How to Run

1. Clone this repository
2. Open `analysis.ipynb`(or whatever file for your final project) in Jupyter or VS Code (open in your local computer if you have Python and Jupyter; otherwise use OpenonDemand Jupyter app)
3. Run all cells from top to bottom

```

**Essential Markdown syntax:**

| What you type | What it renders as |
|---|---|
| `# Heading 1` | Large title |
| `## Heading 2` | Section heading |
| `**bold text**` | **bold text** |
| `` `inline code` `` | `inline code` |
| `- list item` | bullet point |
| blank line between paragraphs | new paragraph |

## Problem 3: Clone Your Repository

**Cloning** downloads a full copy of your repo to your computer — including the complete history of changes. Unlike a ZIP download, a cloned repo stays *connected* to GitHub, so you can sync changes back and forth.

---
### 📌 Using GitHub Desktop

**Step 1 — Sign in to GitHub Desktop**
1. Open GitHub Desktop
2. Go to **File → Options** (Windows) or **GitHub Desktop → Preferences** (Mac)
3. Under **Accounts**, click **Sign In** → sign in with your GitHub credentials

**Step 2 — Clone your repo**
1. Click **File → Clone Repository**
2. Under the **GitHub.com** tab, find your repo in the list
3. Choose a **Local Path** — where the folder will be saved on your computer (e.g. your Documents or Desktop)
4. Click **Clone**

GitHub Desktop will download the repo and open it. You'll see your files in the left panel and the folder path at the top.

**Step 3 — Open the folder**  
Click **Repository → Show in Explorer** (Windows) or **Show in Finder** (Mac) to see the actual folder on your computer.

---
> ⌨️ **Command Line Alternative**
>
> Open your terminal, navigate to where you want the folder, and run:
> ```
> git clone https://github.com/your-username/your-repo-name.git
> ```
> Then open the folder:
> ```
> cd your-repo-name
> ls
> ```

## Problem 4: Add Your Project Files, Commit & Push

This is the core Git workflow you'll use every time you make progress on your project:

> **Add files → Commit → Push**

- **Commit** — a snapshot of your files at a point in time, with a message describing what changed
- **Push** — sends your local commits up to GitHub so they're saved online

---
### 📌 The workflow, step by step

**Step 1 — Copy your files into the repo folder**  
Using your file browser (Finder / Explorer), copy your project files into the cloned folder you created in Problem 3:
- Your notebook: `analysis.ipynb` (or whatever you named it)
- Your dataset: `data.csv` (or whatever your CSV is called)

**Step 2 — GitHub Desktop detects the changes**  
Switch back to GitHub Desktop. You'll see your new files listed on the left under **Changes** with a green **+** badge — GitHub Desktop noticed new files appeared.

**Step 3 — Write a commit message and commit**
1. At the bottom-left of GitHub Desktop, fill in the **Summary** box:  
   e.g. `"Add project notebook and dataset"`
2. Optionally add a longer description in the **Description** box
3. Click **Commit to main**

**Step 4 — Push to GitHub**  
Click **Push origin** (the button at the top-right). This uploads your commit to GitHub.

**Step 5 — Verify on GitHub**  
Go to your repo URL in the browser — your files should now appear.

---
> ⌨️ **Command Line Alternative**
>
> After copying your files into the repo folder, open your terminal inside that folder and run:
> ```
> git add .
> git commit -m "Add project notebook and dataset"
> git push
> ```
> - `git add .` — stages *all* changed files
> - `git commit -m "..."` — creates the snapshot with a message
> - `git push` — sends it to GitHub

## 📌 How Collaboration Works on GitHub

In this lab, **everyone created their own individual repo** to practice the full workflow. But when working on a real group project, the setup is different — the whole team shares a single repo.

---
### The group project model

```
One person creates the repo
        ↓
Add all teammates as collaborators (Settings → Collaborators)
        ↓
Everyone clones the SAME repo to their own computer
        ↓
Each person works on their part locally
        ↓
Everyone commits & pushes their changes to the shared repo
        ↓
⚠️ Before pushing: always Pull first to get your teammates' latest changes
```

**The golden rule of shared repos:**

> **Pull before you push.**  
> Before you start working and before you push, always click **Fetch / Pull origin** in GitHub Desktop (or run `git pull`) to download any changes your teammates pushed since you last synced. This avoids conflicts.

---
### Roles in a shared repo

| Person | What they do |
|---|---|
| **Repo owner** | Creates the repo, adds collaborators, manages settings |
| **Collaborators** | Clone the repo, make changes, push directly to it |
| **Public viewers** | Can see and clone the repo, but cannot push |

---
> 🚨 **For your final project:**  
> Your group will work from **one shared repo** — only one person creates it and adds the rest as collaborators. Everyone clones that same repo and pushes their work there.  
> **Do not create separate repos per person for the group project.**

## Wrap-Up: The Core GitHub Workflow

You now know the complete beginner GitHub loop:

```
Create repo → Write README → Clone to computer
     → Add files → Commit → Push → Share
```

**Commands / actions you used today:**

| Action | GitHub Desktop | Command Line |
|---|---|---|
| Copy repo to your computer | File → Clone Repository | `git clone <url>` |
| Stage changed files | Automatic (checkbox) | `git add .` |
| Save a snapshot with a message | Commit to main | `git commit -m "message"` |
| Upload to GitHub | Push origin | `git push` |
| Download latest changes | Fetch / Pull origin | `git pull` |


**Everytime you want to make a change:**

- First pull any changes
- Stage changes (git add .)
- Commit changes (git commit -m "message")
- Push to github
